```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
from shapely.geometry import Point

# -----------------------------
# Load facility data
# -----------------------------
df = pd.read_csv("data/texas_obs_facilities_final.csv")

# Remove missing coordinates
df = df.dropna(subset=["LATITUDE", "LONGITUDE"])

# Convert to GeoDataFrame
geometry = gpd.points_from_xy(df["LONGITUDE"],
                              df["LATITUDE"])

gdf_fac = gpd.GeoDataFrame(
    df,
    geometry=geometry,
    crs="EPSG:4326"
)

# -----------------------------
# Load Texas county shapefile
# -----------------------------
# Download Texas county shapefile from Census TIGER (https://www2.census.gov/geo/tiger/TIGER2024/COUNTY/?utm_source=chatgpt.com)
counties = gpd.read_file(
    "data/tl_2024_us_county/tl_2024_us_county.shp"
)

# Keep only Texas counties (STATEFP=48)
tx_counties = counties[counties["STATEFP"] == "48"].copy()

# County name consistency
tx_counties["COUNTY"] = (
    tx_counties["NAME"]
    .str.upper()
    .str.replace(" COUNTY", "", regex=False)
)

facility_counties = (
    df["COUNTY"]
    .str.upper()
    .unique()
)

# Counties with no facilities
tx_counties["HasFacility"] = (
    tx_counties["COUNTY"].isin(facility_counties)
)

# -----------------------------
# Plot
# -----------------------------
fig, ax = plt.subplots(figsize=(12, 10))

# Counties WITH facilities
tx_counties[
    tx_counties["HasFacility"]
].plot(
    ax=ax,
    color="lightgray",
    edgecolor="black",
    linewidth=0.4
)

# Counties WITHOUT facilities
tx_counties[
    ~tx_counties["HasFacility"]
].plot(
    ax=ax,
    color="red",
    alpha=0.6,
    edgecolor="black",
    linewidth=0.4,
    label="No Maternity Facility"
)

# Plot facility locations
gdf_fac.plot(
    ax=ax,
    marker="^",
    color="blue",
    markersize=18,
    label="Maternity Facility"
)

ax.set_title(
    "Texas Maternity Care Facilities and Counties Without Facilities",
    fontsize=16
)

ax.set_axis_off()
ax.legend()

plt.tight_layout()
plt.savefig
plt.show()
```
